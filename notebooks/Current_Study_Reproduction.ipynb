# Current study reproduction

Pass 1: 367 topics, 43 supplied finer categories, ten overlapping broad groups. Independent SMDI validation is pending. This notebook calls the current scripts; it does not duplicate their calculations.

**Default: verify aggregate results only.** No scraping, model fit, private-data upload or human annotation is performed. To use Colab, place a clone or extracted companion in your session and set REPO_ROOT below. Install `requirements-domain.txt` into the selected interpreter if required. Exact dependency versions are recorded, but full source reconstruction needs restricted inputs.


```python
from pathlib import Path
import os, sys, subprocess, json, csv

# Optional overrides: repository directory and an existing Python interpreter.
REPO_ROOT = os.environ.get("DUBAI_RESEARCH_ROOT", "")
PYTHON = os.environ.get("DUBAI_RESEARCH_PYTHON", sys.executable)
relative = Path("analysis/submission_2026_09_09/final_corpus_refit")
candidates = [Path(REPO_ROOT).expanduser()] if REPO_ROOT else [Path.cwd(), *Path.cwd().parents]
ROOT = next((p.resolve() for p in candidates if (p / relative / "run_pipeline.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Set REPO_ROOT to the repository or extracted companion root.")
ANALYSIS = ROOT / relative
RESULTS = ANALYSIS / "results/domain_graphs"
print("Repository:", ROOT)
print("Interpreter:", PYTHON)
print("Default operation: aggregate verification; no model refit")
```

## Verify the saved results

This checks denominators, complete-week counts, 96 main specification/lag rows and 72 additional weighted/filtered rows. It cannot establish whether category assignments are semantically correct.


```python
subprocess.run([PYTHON, "-X", "utf8", str(ANALYSIS / "run_pipeline.py"), "--verify"], cwd=ROOT, check=True)
```

## Inspect current weekly data and definitions

`share_all` uses all retained platform records, including topic -1, as the denominator. Groups overlap. DLD outcomes are unique sales registrations and their recorded amounts, not a constant-quality price index. The paired plots and explanations distinguish primary, exploratory and contextual comparisons.


```python
for name in ["domain_weekly.csv", "dld_weekly.csv", "primary_contemporaneous.csv"]:
    with (RESULTS / name).open(encoding="utf8", newline="") as stream:
        rows = list(csv.DictReader(stream))
    print("\n", name, "rows:", len(rows))
    for row in rows[:2]: print(row)
print("\nPlot explanations:", RESULTS / "comparisons/GRAPH_EXPLANATIONS.md")
print("Comparison PDF:", RESULTS / "comparisons/domain_market_comparisons.pdf")
print("Semantic codebook:", ROOT / "paper/semantic_codebook.pdf")
```

```python
REBUILD_FROM_FROZEN_INPUTS = False
if REBUILD_FROM_FROZEN_INPUTS:
    subprocess.run([PYTHON, "-X", "utf8", str(ANALYSIS / "run_pipeline.py"), "--check-inputs"], cwd=ROOT, check=True)
    subprocess.run([PYTHON, "-X", "utf8", str(ANALYSIS / "run_pipeline.py")], cwd=ROOT, check=True)
else:
    print("Rebuild skipped. Saved aggregate results were verified above.")
```
